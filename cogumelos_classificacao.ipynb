#### Contextualização do Problema
Este projeto aborda o clássico problema de classificação binária para determinar se um cogumelo é **comestível** (edible, representado por 'e') ou **venenoso** (poisonous, representado por 'p'), com base em suas características morfológicas e físicas. 
O dataset original foi criado pelo *Audubon Society Field Guide to North American Mushrooms* (1981) e contém descrições hipotéticas de amostras correspondentes a 23 espécies de cogumelos branquiais da família Agaricus e Lepiota.

A classificação precisa é vital na prática de coleta de cogumelos silvestres, pois a ingestão de cogumelos venenosos pode causar graves problemas de saúde, inclusive a morte. Este estudo demonstra como modelos de aprendizado de máquina podem identificar padrões e auxiliar nessa classificação com alta precisão.

### 4. Tratamento dos Dados
Nesta etapa, investiguei valores nulos/ausentes e defini a melhor estratégia para tratá-los. Também tratei de variáveis constantes que não agregam informação e realizei a codificação (encoding) das variáveis qualitativas para sua representação numérica.

```python
# 1. Identificação de valores ausentes explícitos (nulos padrão do Pandas)
print("Valores nulos identificados por isna().sum():")
print(df.isna().sum())
```

```python
# 2. Investigação de valores ausentes implícitos representados pelo caractere '?'
# De acordo com a documentação do dataset, a variável 'stalk-root' possui valores ausentes representados por '?'
missing_stalk_root = (df['stalk-root'] == '?').sum()
print(f"Quantidade de valores '?' na coluna 'stalk-root': {missing_stalk_root}")
print(f"Porcentagem de dados ausentes em 'stalk-root': {(missing_stalk_root / len(df)) * 100:.2f}%")
```

**Decisão e Justificativa sobre Valores Ausentes:**
A coluna `stalk-root` contém **2.480 registros** com o valor `'?'` (aproximadamente **30.53%** do total do dataset).

Estratégias avaliadas:
1. **Excluir os registros:** Perderia mais de 30% da minha base de dados, reduzindo severamente o volume de treinamento e o poder preditivo do modelo.
2. **Excluir a coluna inteira:** Poderia perder informações cruciais sobre a morfologia da raiz do talo.
3. **Imputar os dados pela moda (classe mais frequente):** A moda da coluna é `b` (bulbosa). Imputar 30% dos dados com um único valor pode introduzir forte viés e distorcer a realidade biológica dos cogumelos.
4. **Criar uma nova categoria ("missing" ou "unknown"):** Como se trata de variáveis categóricas puras, transformei o `'?'` em uma categoria explícita chamada `'missing'`. Desta forma, mantenho todos os registros e permito que o modelo aprenda se a *ausência* ou a *dificuldade de observar* a raiz do talo está correlacionada com a toxicidade do cogumelo.

**Ação:** Adotei a estratégia 4 (criar a categoria `'missing'`).

```python
# Substituição do caractere '?' por 'missing' na coluna stalk-root
df['stalk-root'] = df['stalk-root'].replace('?', 'missing')

# Remoção da coluna 'veil-type', pois ela possui um único valor constante ('p') e não contribui para o modelo
df = df.drop(columns=['veil-type'])

print("Colunas restantes após o tratamento:", list(df.columns))
print(f"Novo formato do DataFrame: {df.shape}")
```

### 6. Preparação dos Dados
Nesta etapa, preparei os dados para os algoritmos de Machine Learning:
1. **Separação de X (atributos) e y (variável alvo).**
2. **Codificação da Variável Alvo (`class`):** Usei o `LabelEncoder` para transformar as classes `e` (comestível) e `p` (venenoso) em `0` e `1`.
3. **Codificação dos Atributos Preditoras (`X`):** Utilizei a técnica de **One-Hot Encoding** (`pd.get_dummies()`). Essa técnica cria colunas binárias (0 ou 1) para cada categoria de cada atributo preditor. Isso é essencial porque modelos como KNN e redes neurais dependem de distâncias matemáticas numéricas para funcionar corretamente.
4. **Divisão de Treino e Teste:** Dividi os dados em 70% para treinamento e 30% para teste, garantindo a estratificação (`stratify`) com base na proporção das classes originais.

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

# 1. Divisão entre Atributos (X) e Variável Alvo (y)
X = df.drop(columns=['class'])
y = df['class']

# 2. Codificação da variável alvo (y) com LabelEncoder
le = LabelEncoder()
y_encoded = le.fit_transform(y)
# e (comestível) -> 0, p (venenoso) -> 1

# Exibição do mapeamento das classes
print("Mapeamento do LabelEncoder:")
for i, classe in enumerate(le.classes_):
    print(f"Classe original: '{classe}' -> Codificação numérica: {i}")

# 3. Codificação dos atributos preditores (X) com One-Hot Encoding
X_encoded = pd.get_dummies(X, drop_first=False)

print(f"\nShape original de X: {X.shape}")
print(f"Shape de X após One-Hot Encoding: {X_encoded.shape}")
```

```python
# 4. Divisão em dados de treino (70%) e teste (30%)
# O parâmetro 'stratify=y_encoded' garante que as proporções de comestível e venenoso sejam mantidas em treino e teste.
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y_encoded, test_size=0.3, stratify=y_encoded, random_state=42
)

print("Divisão concluída:")
print(f"Conjunto de Treino: {X_train.shape[0]} amostras (70%)")
print(f"Conjunto de Teste: {X_test.shape[0]} amostras (30%)")
```

### 7. Construção do Modelo de Classificação
Para resolver este problema de classificação, construí e treinei três classificadores diferentes:
1. **Decision Tree (Árvore de Decisão):** Modelo obrigatório e altamente interpretável que divide o espaço de busca com base em entropia ou ganho de informação (índice Gini).
2. **Random Forest (Floresta Aleatória):** Um modelo baseado em *ensemble bagging* que combina várias árvores de decisão para reduzir a variância e mitigar o overfitting.
3. **K-Nearest Neighbors (KNN):** Modelo baseado em instâncias que calcula a proximidade dos vizinhos mais próximos para inferir a classe da amostra.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

# Instanciação dos classificadores
dt_model = DecisionTreeClassifier(random_state=42)
rf_model = RandomForestClassifier(random_state=42)
knn_model = KNeighborsClassifier(n_neighbors=5)

# Treinamento dos modelos com os dados de treino
print("Treinando Decision Tree Classifier...")
dt_model.fit(X_train, y_train)

print("Treinando Random Forest Classifier...")
rf_model.fit(X_train, y_train)

print("Treinando KNN Classifier...")
knn_model.fit(X_train, y_train)

print("\nTreinamento de todos os modelos finalizado com sucesso!")

from sklearn.model_selection import cross_val_score

print("\n=== Cross Validation ===")

# Decision Tree
dt_scores = cross_val_score(dt_model, X_encoded, y_encoded, cv=5)
print("Decision Tree:", dt_scores.mean())

# Random Forest
rf_scores = cross_val_score(rf_model, X_encoded, y_encoded, cv=5)
print("Random Forest:", rf_scores.mean())

# KNN
knn_scores = cross_val_score(knn_model, X_encoded, y_encoded, cv=5)
print("KNN:", knn_scores.mean())
```

### 8. Avaliação do Modelo
A avaliação será feita utilizando três métricas/ferramentas principais para o conjunto de testes:
- **Acurácia:** Proporção geral de predições corretas.
- **Classification Report:** Exibe métricas detalhadas como Precision (Precisão), Recall (Revocação/Sensibilidade) e F1-Score para cada classe.
- **Matriz de Confusão:** Tabela que cruza os valores reais e as predições do modelo, permitindo visualizar de forma clara onde ocorreram os erros.

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Dicionário com os modelos treinados para facilitar a iteração
models = {
    "Decision Tree": dt_model,
    "Random Forest": rf_model,
    "KNN (k=5)": knn_model
}

# Dicionário para salvar as acurácias
accuracies = {}
cv_accuracies = {}

# Avaliação individual de cada modelo
for name, model in models.items():
    print(f"\n==================== AVALIAÇÃO: {name.upper()} ====================")
    
    # Realizando predições no conjunto de teste
    y_pred = model.predict(X_test)
    
    # Calculando acurácia
    acc = accuracy_score(y_test, y_pred)
    accuracies[name] = acc
    # Calculando validação cruzada (5-fold)
    cv_score = cross_val_score(model, X_encoded, y_encoded, cv=5).mean()
    cv_accuracies[name] = cv_score
    print(f"Acurácia no conjunto de teste (Holdout): {acc * 100:.2f}%")
    print(f"Acurácia Média na Validação Cruzada (5-fold CV): {cv_score * 100:.2f}%")
    
    # Calculando Matriz de Confusão e Classification Report
    cm = confusion_matrix(y_test, y_pred)
    cr = classification_report(y_test, y_pred, target_names=["Comestível (e)", "Venenoso (p)"])
    
    print("\nClassification Report:")
    print(cr)
    
    # Plot da Matriz de Confusão usando Seaborn
    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=["e (Comestível)", "p (Venenoso)"], yticklabels=["e (Comestível)", "p (Venenoso)"], cbar=False)
    plt.title(f"Matriz de Confusão - {name}", fontsize=12, pad=10)
    plt.ylabel("Classe Real", fontsize=10)
    plt.xlabel("Classe Predita", fontsize=10)
    plt.tight_layout()
    plt.show()
```

```python
# Tabela Comparativa de Desempenho
df_comparacao = pd.DataFrame({
    'Modelo': list(accuracies.keys()),
    'Acurácia Teste (%)': [v * 100 for v in accuracies.values()],
    'Acurácia CV (%)': [v * 100 for v in cv_accuracies.values()]
})
df_comparacao = df_comparacao.sort_values(by='Acurácia CV (%)', ascending=False)

print("Tabela Comparativa de Acurácia:")
df_comparacao.reset_index(drop=True)
```

A estrutura lógica das regras de sobrevivência biológica dos cogumelos (por exemplo: "se cheira mal ou tem esporos chocolate, é venenoso") é perfeitamente capturada pelo critério de partição das árvores de decisão.

### 10. Teste de Predição com Novos Cogumelos
Insira as características de um cogumelo no dicionário abaixo e execute a célula para que a Inteligência Artificial classifique-o como **Comestível** ou **Venenoso**.

```python
# Dicionário com as características do cogumelo para classificar
# Altere os valores abaixo para testar diferentes características
novo_cogumelo = {
    'cap-shape': 'x',              # formato do chapéu: b=sino, c=cônico, x=convexo, f=plano, k=nodoso, s=afundado
    'cap-surface': 's',            # superfície: f=fibroso, g=sulcos, y=escamoso, s=liso
    'cap-color': 'n',              # cor do chapéu: n=marrom, b=amarelo, g=cinza, etc.
    'bruises': 't',                # manchas/hematomas: t=tem, f=não tem
    'odor': 'n',                   # odor: a=amêndoa, l=anis, c=creosoto, y=peixe, f=fétido, m=mofo, n=nenhum, p=pungente, s=picante
    'gill-attachment': 'f',        # inserção das lamelas: a=conectada, f=livre
    'gill-spacing': 'c',           # espaçamento das lamelas: c=fechado, w=largo
    'gill-size': 'b',              # tamanho das lamelas: b=largo, n=estreito
    'gill-color': 'k',             # cor das lamelas: k=preto, n=marrom, g=cinza, etc.
    'stalk-shape': 'e',            # formato do talo: e=alargado, t=afunilado
    'stalk-root': 'b',             # raiz do talo: b=bulbosa, c=clava, e=igual, r=enraizada, missing=ausente
    'stalk-surface-above-ring': 's',# superfície do talo acima do anel: f, y, k, s
    'stalk-surface-below-ring': 's',# superfície do talo abaixo do anel: f, y, k, s
    'stalk-color-above-ring': 'w', # cor do talo acima do anel: n, b, c, g, o, p, e, w, y
    'stalk-color-below-ring': 'w', # cor do talo abaixo do anel: n, b, c, g, o, p, e, w, y
    'veil-color': 'w',             # cor do véu: n, o, w, y
    'ring-number': 'o',            # número de anéis: n=nenhum, o=um, t=dois
    'ring-type': 'p',              # tipo de anel: e, f, l, n, p
    'spore-print-color': 'n',      # cor do esporo: k=preto, n=marrom, h=chocolate, w=branco, r=verde, etc.
    'population': 'v',             # população: a, c, n, s, v, y
    'habitat': 'd'                 # habitat: g, l, m, p, u, w, d
}

def classificar_cogumelo(dados, modelo, colunas_treino):
    # 1. Filtro de Segurança Biológico (Regra de Especialista)
    # Na natureza e no dataset, cogumelos com odores nocivos (f, y, c, p, s, m) são sempre venenosos.
    odores_venenosos = ['f', 'y', 'c', 'p', 's', 'm']
    if dados.get('odor') in odores_venenosos:
        print("=" * 60)
        print("              RESULTADO DA CLASSIFICAÇÃO DA IA              ")
        print("=" * 60)
        print(" Diagnóstico: O cogumelo é VENENOSO ⚠️")
        print(" Confiança (Filtro de Segurança): 100.00%")
        print(" Observação: Classificado via Filtro de Segurança Biológico (Odor Nocivo).")
        print("=" * 60)
        return
    # Criar DataFrame
    df_novo = pd.DataFrame([dados])
    
    # Tratar nulos
    if 'stalk-root' in df_novo.columns:
        df_novo['stalk-root'] = df_novo['stalk-root'].replace('?', 'missing')
    if 'veil-type' in df_novo.columns:
        df_novo = df_novo.drop(columns=['veil-type'])
        
    # One-hot encoding
    df_novo_encoded = pd.get_dummies(df_novo)
    
    # Criar base com colunas corretas
    df_final = pd.DataFrame(0, index=[0], columns=colunas_treino)
    for col in df_novo_encoded.columns:
        if col in df_final.columns:
            df_final[col] = df_novo_encoded[col].values[0]
            
    # Predição
    classe_predita = modelo.predict(df_final)[0]
    probabilidade = modelo.predict_proba(df_final)[0]
    
    # Mapeamento do LabelEncoder para exibir texto legível
    resultado = "VENENOSO ⚠️" if classe_predita == 1 else "COMESTÍVEL ✅"
    confianca = probabilidade[classe_predita] * 100
    
    print("=" * 60)
    print("              RESULTADO DA CLASSIFICAÇÃO DA IA              ")
    print("=" * 60)
    print(f" Diagnóstico: O cogumelo é {resultado}")
    print(f" Confiança do Modelo: {confianca:.2f}%")
    print("=" * 60)

# Classifica usando o modelo de Árvore de Decisão treinado anteriormente
classificar_cogumelo(novo_cogumelo, dt_model, X_train.columns)
```
